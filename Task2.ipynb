```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('/content/Telco_Cusomer_Churn.csv')

# 2. Structural Data Type Cleaning
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].str.strip(), errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)

# --- ADVANCED BUSINESS KPI EXTRACTION ---

# Total Revenue Leakage Metrics
total_active_subscribers = df[df['Churn'] == 'No'].shape[0]
total_churned_subscribers = df[df['Churn'] == 'Yes'].shape[0]
global_churn_rate = (total_churned_subscribers / len(df)) * 100

monthly_revenue_at_risk = df[df['Churn'] == 'Yes']['MonthlyCharges'].sum()
historical_lost_revenue = df[df['Churn'] == 'Yes']['TotalCharges'].sum()

# Print out a clean, high-impact Executive Health Check Table
print("="*70)
print("             🚨 EXECUTIVE REVENUE & RETENTION AUDIT REPORT 🚨             ")
print("="*70)
print(f"📊 CUSTOMER BASEMETRICS:")
print(f"  • Total Customers Tracked       : {len(df):,}")
print(f"  • Active Retained Subscribers   : {total_active_subscribers:,} ({100 - global_churn_rate:.1f}%)")
print(f"  • Terminated (Churned) Accounts : {total_churned_subscribers:,} ({global_churn_rate:.1f}%)")
print("-"*70)
print(f"💰 FINANCIAL IMPACT MATRIX:")
print(f"  • Total Historical Revenue Lost : ${historical_lost_revenue:,.2f}")
print(f"  • Monthly Recurring Revenue (MRR) Lost Today: ${monthly_revenue_at_risk:,.2f}")
print("="*70)

# Display structural dataset summaries below the financials
print("\n📋 DATA INFRASTRUCTURE HEALTH:")
print(f"• Shape: {df.shape[0]} rows | {df.shape[1]} columns")
print(f"• Missing data points resolved: {df['TotalCharges'].isnull().sum()}")
print(f"• Duplicated entities dropped  : {df.duplicated().sum()}")
print("-"*70)
df.head(3)
```

### 🛠️ Data Integrity & Transformation Methodology
To prevent metric distortion during downstream retention visualization, we initiated an explicit cleaning protocol:
* **Datatype Coercion:** The feature `TotalCharges` was natively cataloged as a generic string object due to whitespace omissions. These profiles were programmatically forced into an analytical `float64` framework.
* **Preservation Imputation:** 11 active accounts showcased null value bounds in their lifetime pricing tallies. Investigation verified these accounts held a `tenure` of 0 months. Rather than dropping valid customer matrices, these boundaries were imputed to `$0.00` to reflect zero billing history.
